### Delhi Air Quality Dataset (2025)

**File:** `delhi_merged_2025_1.csv`

#### About this dataset

This dataset records **air pollution and weather conditions in Delhi, India, hour by hour, for the whole of 2025**. Each row is one reading from one monitoring station at one hour. It can be used to see how pollution changes across the year, across the day, from one part of the city to another, and with the weather. Delhi is known for high air pollution, especially in winter, so this is the kind of data used to study it.

The data was put together by merging readings from **three different monitoring networks** into a single table. Because the networks use different instruments and measure different things, not every station reports every column. This is the main reason many columns contain a lot of empty values.

#### Quick facts

- **Period:** Jan 1, 2025 00:00 to Dec 31, 2025 23:00 (all 12 months)
- **Location:** Delhi, with 46 station IDs spread across the city (45 distinct names)
- **Size:** 395,664 rows × 32 columns
- **Structure:** One row per station per hour. A complete station has 8,760 rows (24 hours × 365 days).
- **Missing values:** Empty cells mean no reading was recorded. They are stored as blanks (NaN), not as zero.

#### Where the data comes from

- **DPCC** (Delhi Pollution Control Committee): 30 stations, the largest network
- **IITM** (Indian Institute of Tropical Meteorology): 10 stations
- **CPCB** (Central Pollution Control Board): 6 stations

Each station belongs to exactly one network. As a rough guide, CPCB and DPCC stations report most pollutants, while IITM stations mostly report particles and gases but few weather columns.

#### Known gaps in station coverage

- **Six stations have no measurements at all.** Jnu, Nsut Jaffarpur, Cantonment Area, Commonwealth Sports Complex, Ignou Maidan Garhi and Talkatora Garden (IDs 6076 to 6081, all DPCC) were added to the system in 2025 while still under construction. They only opened in 2026, so they have no data for 2025. Their 52,560 rows (13% of the file) are placeholders.
- **`TOT-RF` is filled with 0.0 for those six stations**, so it looks almost complete (7% missing) but those zeros are not real readings.
- **Iit Delhi (ID 6060) is partial.** It has only 1,464 rows, covering Nov and Dec, and about half of those hours contain data.
- **"Pusa" appears twice.** IDs 107 (IITM) and 1563 (DPCC) are different stations that share a name, so use `station_id` to tell them apart.

#### How to read the column tables

The columns are split into four groups. The **Missing** column shows the share of empty cells over all 395,664 rows, including the six empty stations.

```python
def frozen_repeats(df, col, min_len=6):

    run_id = (
        df.groupby('station_id')[col]
          .transform(lambda x: x.ne(x.shift()).cumsum())
    )

    key = (
        df['station_id'].astype(str)
        + '_'
        + run_id.astype(str)
    )

    run_len = df.groupby(key)[col].transform('size')

    is_first = ~key.duplicated()

    mask = (
        (run_len >= min_len)
        & df[col].notna()
        & ~is_first
    )

    return mask
```

```python
# 1. impossible values
#-----i have decided not to null the zeros because realisticaly think value can exist they arn not impossible and good amount of them exist during monsoon 
#

# 2. stuck sensor: same value for more than 6 hours in a row at one station
new_run = (df["CO (mg/m³)"] != df["CO (mg/m³)"].shift()) | (df["station_id"] != df["station_id"].shift()) | df["CO (mg/m³)"].isna()
run_len = df["CO (mg/m³)"].groupby(new_run.cumsum()).transform("size").where(df["CO (mg/m³)"].notna(), 0)

stuck = run_len > 6
print("CO stuck values set to NaN:", int(stuck.sum()))
df.loc[stuck, "CO (mg/m³)"] = np.nan



# CO: Burari Crossing, 26 Jul 21:00 to 27 Jul 14:00 -> instrument fault, CO set to missing (rows kept)
#
# WHY THESE 16 READINGS ARE REMOVED:
#   1. Sudden jump: CO went from about 1 mg/m³ (20:00) to 30 mg/m³ (21:00), then stayed
#      flat between 46 and 49 mg/m³ for 15 hours. Real CO rises and falls with traffic and
#      burning; a flat line near the top of the sensor's range (about 50 mg/m³) means a stuck sensor.
#   2. Far outside this station's normal: in its other 8,181 hours, Burari Crossing has a
#      median of 0.91 mg/m³ and a maximum of 7.79 mg/m³.
#   3. Far outside the whole dataset: the highest CO anywhere else is 9.87 mg/m³, so there is
#      nothing between 9.87 and 30. These readings are 3 to 5 times larger than any other.
#   4. No other station saw it: in the same hours the other 35 CO stations had a median of
#      0.74 and a maximum of 3.66 mg/m³. CO spreads across a city, so a real event would show nearby.
#   5. Other pollutants did not react: PM2.5 at this station was lower than usual (median 25.7
#      against 63.1), so there was no smoke or fire to explain it.
#   6. Not physically believable: 48 mg/m³ is about 12 times India's 1-hour CO limit (4 mg/m³).
#
# We only blank the CO value. PM2.5, NO2 and the other columns for these hours look normal, so we keep the rows.

fault = (
    (df["station_name"] == "Burari Crossing")
    & (df["Timestamp"] >= "2025-07-26 21:00")
    & (df["Timestamp"] <= "2025-07-27 14:00")
)
print("CO values set to NaN:", int((fault & df["CO (mg/m³)"].notna()).sum()))
df.loc[fault, "CO (mg/m³)"] = np.nan
```

```python


# 1. more than 40 points away from the city median at the same hour
city_median = df.groupby("Timestamp")["RH (%)"].transform("median")
off = (df["RH (%)"] - city_median).abs() > 40
print("RH far from city median set to NaN:", int(off.sum()))
df.loc[off, "RH (%)"] = np.nan

# 2. stuck sensor: same value for more than 12 hours in a row at one station
new_run = (df["RH (%)"] != df["RH (%)"].shift()) | (df["station_id"] != df["station_id"].shift()) | df["RH (%)"].isna()
run_len = df["RH (%)"].groupby(new_run.cumsum()).transform("size").where(df["RH (%)"].notna(), 0)
stuck = run_len > 12
print("RH stuck values set to NaN:", int(stuck.sum()))
df.loc[stuck, "RH (%)"] = np.nan

# RH: a station whose typical (median) humidity over a whole month is more than 20 points away
# from the city median at the same hours is not trustworthy that month (normal stations stay within 12).
city_median = df.groupby("Timestamp")["RH (%)"].transform("median")
gap = df["RH (%)"] - city_median
month = df["Timestamp"].dt.month

monthly_gap = gap.groupby([df["station_id"], month]).transform("median")      # typical gap of this station in this month
n_month = df["RH (%)"].groupby([df["station_id"], month]).transform("count")   # readings in that station-month

bad_month = (monthly_gap.abs() > 20) & (n_month >= 200)    # only judge months with enough readings
print("RH values set to NaN in unreliable station-months:", int((bad_month & df["RH (%)"].notna()).sum()))
df.loc[bad_month, "RH (%)"] = np.nan
```

```python

# SOLAR RADIATION (SR) CLEANING
# SR is measured in W/m². It should be close to 0 at night and peak around noon.

# Step 1: mark which hours are night (8 pm to 4:59 am).
# At night there is no sunlight, so SR should be near zero.
df["Timestamp"] = pd.to_datetime(df["Timestamp"])      # make sure Timestamp is a date column
hour = df["Timestamp"].dt.hour
night = (hour >= 20) | (hour <= 4)

# Step 2: remove the four stations whose SR sensors are faulty.
#   Lodhi Road            -> reads about 220 all day and night (dead sensor)
#   Shadipur              -> 220 at midnight, values up to 1,878, frozen readings
#   IHBAS Dilshad Garden  -> very low readings, then a jump in June (dirty or covered sensor)
#   Sonia Vihar           -> weak, flat signal that never rises like real sunlight
faulty_stations = ["Lodhi Road", "Shadipur", "IHBAS Dilshad Garden", "Sonia Vihar"]
n_removed = int(df.loc[df["station_name"].isin(faulty_stations), "SR (W/mt2)"].notna().sum())
print("SR values removed at faulty stations:", n_removed)
df.loc[df["station_name"].isin(faulty_stations), "SR (W/mt2)"] = np.nan

# Step 3: at the remaining stations, remove readings above 50 W/m² at night.
# Real sunlight can't be that strong after dark, so these values are errors.
n_night = int((night & (df["SR (W/mt2)"] > 50)).sum())
print("SR night values above 50 removed:", n_night)
df.loc[night & (df["SR (W/mt2)"] > 50), "SR (W/mt2)"] = np.nan
```

```python
#  Atmospheric Pressure: rename and fix units
df = df.rename(columns={"BP (mmHg)": "BP (hPa)"})

# 1. Anand Vihar reports in mmHg: convert to hPa (1 mmHg = 1.33322 hPa)
anand = df["station_name"] == "Anand Vihar"
df.loc[anand, "BP (hPa)"] = df.loc[anand, "BP (hPa)"] * 1.33322

# 2. NSIT Dwarka reads about 34 hPa too low even after converting: faulty sensor, blank it
df.loc[df["station_name"] == "NSIT Dwarka", "BP (hPa)"] = np.nan

# 3. Impossible values for Delhi (normal range is about 960 to 1000 hPa)
bad_range = df["BP (hPa)"].notna() & ~df["BP (hPa)"].between(930, 1010)
print("BP outside 930-1010 hPa set to NaN:", int(bad_range.sum()))
df.loc[bad_range, "BP (hPa)"] = np.nan

# 4. Stuck sensor: same value for more than 72 hours in a row at one station
new_run = (df["BP (hPa)"] != df["BP (hPa)"].shift()) | (df["station_id"] != df["station_id"].shift()) | df["BP (hPa)"].isna()
run_len = df["BP (hPa)"].groupby(new_run.cumsum()).transform("size").where(df["BP (hPa)"].notna(), 0)
stuck = run_len > 72
print("BP stuck values set to NaN:", int(stuck.sum()))
df.loc[stuck, "BP (hPa)"] = np.nan
```
